```python
import pandas as pd

# Load the Excel file
file_path = 'C:/Users/Ross/Dropbox/GABM Epidemic Round #2/Analysis/Collective Analysis/Full Feedback Figure 4 Data/error-rate.xlsx'
df = pd.read_excel(file_path)

# Focus on the specified columns and rows (timesteps 0 through 85, rows 2 to 1001)
df_timesteps = df.loc[1:1000, '0':'85']

# Convert all values to strings to differentiate numeric '1' and '0' from string responses
df_timesteps_str = df_timesteps.astype(str)

# Replace '1.0' and '0.0' (which are now strings) with an empty string
df_timesteps_str.replace(['1.0', '0.0'], '', inplace=True)

# Filter out numeric responses and '1.0', '0.0' (considered as string after conversion)
non_numeric_responses = df_timesteps_str.applymap(
    lambda x: x if (not x.isnumeric() and x not in ['1.0', '0.0', '']) else None)

# Count the frequency of each unique non-numeric value (including repeats)
non_numeric_value_counts = non_numeric_responses.stack().value_counts()

# Sum up the total number of non-numeric, non-binary responses
total_non_numeric_responses = non_numeric_value_counts.sum()

non_numeric_value_counts, total_non_numeric_responses
```
